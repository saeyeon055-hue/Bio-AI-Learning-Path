# 1. Class
> 클래스: 자료와 그 자료에 대한 기능(함수)를 묶어 높은 프로그래밍 단위
- 속성(attribute), 클래스 변수(class variable): 클래스에 정의된 자료

- 메소드(method): 클래스에 정의된 함수

> 클래스를 메모리에 할당하여 사용하면 인스턴스(객체)가 됨
- 클래스는 추상적인 개념이며 인스턴스는 그것을 구체화하여 실현한 것

````python
class Calc:

  my_name = "I'm Calc Module!"

  def add(self, a, b):
    result = a + b
    print("{0:f} + {1:f} = {2:f}".format(a,b,result))
  def sub(self, a, b):
    result = a - b
    print("{0:f} - {1:f} = {2:f}".format(a,b,result))
  def mul(self, a, b):
    result = a * b
    print("{0:f} * {1:f} = {2:f}".format(a,b,result))
  def div(self, a, b):
    if b !=0:
      result = a / b
      print("{0:f} / {1:f} = {2:f}".format(a,b,result))
    else:
      print("Divided by zero!")
````

# 2. 생성자 (__init__)

> 클래스가 인스턴스화될 때 자동 수행되는 메소드

> 주로 속성 값 초기화가 필요할 때 재정의

````python
class Calc:

  my_name = "I'm Calc Module!"
  scale = 0 # 생성자에서 정의할 경우, 이곳에 선언해도 되고 안해도 됨.

  def __init__(self,scale=1):
    self.scale = scale

  def add(self, a, b):
    result = (a + b) * self.scale
    print("{0:f} + {1:f} * {2:f} = {3:f}".format(a,b,self.scale,result))
````

````python
num1=Calc()
num1.add(1,2)

# scale에 값 지정
num2=Calc(10)
num2.add(1,2)
````

생성자는 직접 부르는게 아니다.

num1.__init__(10) ----> (X)

클래스 이름으로 생성할때 넣는 것.

Calc(10) ----> (O)

# 3. 소멸자(__del__)
> 인스턴스가 소멸(메모리에서 제거)될 때 자동 수행되는 메소드

> 주로 메모리에서 제거할 것이 있을 때 재정의

> 소멸자는 기본적으로 파라미터를 가지지 않 (def __del__(self))

````python
class Calc:

  my_name = "I'm Calc Module!"
  scale = 0 # 생성자에서 정의할 경우, 이곳에 선언해도 되고 안해도 됨.

  def __init__(self,scale=1):
    self.scale = scale

  def __del__(self):
    self.scale =1
    print("Scale is reset to 1.")

  def add(self, a, b):
    result = (a + b) * self.scale
    print("{0:f} + {1:f} * {2:f} = {3:f}".format(a,b,self.scale,result))
````

````python
num1=Calc()
num1.add(1,2)

# scale에 값 지정
num2=Calc(10)
num2.add(1,2)

del(num2)
````

# 4. 상속(inheritance)

> 객체 지향 언어 : 객체를 기반으로하는 프로그래밍 언어

> 기존 클래스의 일부를 수저하여 사용한느 메커니즘

> 상속을 해주는 클래스를 부모 클래스(parent class), 상속을 받는 클래스를 자식 클래스(child class)라고 함.

> 자식 클래스는 부모의 속성과 메소드를 사용할 수 있고, 기능을 개선하여 수정할 수 있음.
- 실세계 예: 부모의 자동차를 상속 받아 타이어(속성)를 바꾸고 하늘 날기(메소드)를 추가하는 것

- 형식

```
  class 클래스명(상속 클래스명):
    속성
    ...
    def 메소드(self, 인자1, 인자2):
        <수행할 문장1>
        <수행할 문장2>
        ...
```

````python
class ParentCalc:
  def add(self, a, b):
    result = a + b
    print('{0:f} + {1:f} = {2:f}'.format(a,b,result))
  def sub(self, a, b):
    result = a - b
    print('{0:f} - {1:f} = {2:f}'.format(a,b,result))

class ChildCalc(ParentCalc):
    def add(self, *a):
      s = "{0:f}".format(a[0])
      for i in a[1:]:
        s = s + ' + ' + '{0:f}'.format(i)
      result = sum(a)
      print('{0:s} = {1:f}'.format(s,result))

num=ChildCalc()
num.add(1,2,3,4)
num.sub(1,2)
````

# 5. 접근 권한 (access modifier)
> Publie (디폴트)
- 자식에게 모두 상속되며 자식이 아닌 경우도 접근 가능

> Protected (접근 제한하려면 복잡함. 추천 안함)
- 자식에게 상속되며 자식만 접근 가능

- 형식: 속성명이나 메소드명에 '_'를 붙임

> Private
- 자식에게도 상속되지 않으며 본인만 접근 가능

- 형식: 속성명이나 메소드명에 '__'를 붙임

````python
class ParentCalc:
  my_name = 'Public ParentCalc'
  __my_name = 'Private ParentCalc'
  def show_name(self):
    print(self.my_name)
    print(self.__my_name)

class ChildCalc(ParentCalc):
  my_name = 'Public ChildCalc'
  __my_name = 'Private ChildCalc'
  def show_name(self):
    print(super().my_name) # 'super' 는 부모 객체를 의미. 'self'는 나 자신.
    print(self.my_name)
    print(super().__my_name)

num1=ParentCalc()
num1.show_name()
print('')
print(num1.my_name)
print(num1.__my_name)
````

위 오류는
```
print(num1.__my_name)
```
가 'Private ParentCalc' 이기때문에 접근할 수 없어서 오류가 발생한 것.

````python
num1=ParentCalc()
num1.show_name()
print('')
num2=ChildCalc()
num2.show_name()
````

위 오류 또한, 부모 객체에서 'Private ParentCalc'에 해당하는 부분이 있어서 발생한 것.

# 6. 연산자 오버라이딩 (0perator overriding)
> 연산자 오버라이딩
- 인스턴스 간에 연산을 수행할 수 있도록 연산자(+,-,*,/,...)를 재정의하여 사용하는 것
- 클래스 연산자 메소드를 재정의
 1) self 인스턴스 + other 인스턴스
```
__add__(self, other)
```
  2) self 인스턴스 + other 인스턴스

```
__sub__(self, other)
```
  3) self 인스턴스 * other 인스턴스
```
__mul__(self, other)
```
  4) self 인스턴스 / other 인스턴스
```
__truediv__(self, other)
```

````python
class Point:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __add__(self, other):
    self.x += other.x
    self.y += other.y
    return self # 계산된 값을 다른 인스턴스에 할당(=)하기 위해서 자신을 리턴
  def show(self):
    print('({0:d},{1:d})'.format(self.x,self.y))

a = Point(1,2)
a.show()

b = Point(3,4)
b.show()

a = a + b
a.show()
````

계산된 값을 다른 인스턴스에 할당(=)하기 위해서 자신을 리턴

x = a + b

add에 의해서 a + b가 계산되고, 그 값이 a.x와 a.y에 들어있다
이 값을 리턴해줘야함.

# 7. 예외 처리 (exception handling)
> 프로그래밍 중 발생할 수 있는 여러 에러를 모아서 처리하는 메커니즘

> 다양한 예외가 존재
- 0으로 나누는 경우

- 파일이 없는데 open 하는 경우

- 리스트의 범위를 초과하여 인덱싱하는 경우

```
try:
except[발생에러[as 에러메시지변수]]:
      ...
finally:
      ...
```

  1) try: 정상 상황의 코드

  2) except: 에러가 발생하면 except문을 실행. 에러 발생 시 except 문에 미리 정해놓은'[]'에러와 일치할 때만 except문 실험

  3) finally: 에러 발생 여부와 상관없이 무조건 실행


````python
num1 = int(input('num1='))
num2 = int(input('num2='))
try:
  num3 = num1/num2
  print('{0:d} / {1:d} = {2:f}'.format(num1,num2,num3))
except ZeroDivisionError as e:
  print(e)
finally:
  print('The end!')
````

  - 메소드
    - move(dx,dy): 좌상단, 우하단 포인트 좌표를 (dx, dy)만큼 이동

    - show(): 좌상단, 우하단 포인트 좌표 출력

````python
class Point:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def move(self,dx,dy):
    self.x+=dx
    self.y+=dy
  def show(self):
    print('({0:d},{1:d})'.format(self.x,self.y))

class Box(Point):
  def __init__(self, p1, p2):
    self.p1=p1
    self.p2=p2
  def move(self,dx,dy):
    self.p1.move(dx,dy)
    self.p2.move(dx,dy)
  def show(self):
    self.p1.show()
    self.p2.show()

p1 = Point(1,2)
p1.show()
p2 = Point(3,4)
p2.show()

b = Box(p1,p2)
b.move(1,1)
b.show()
````
